```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.legend_handler import HandlerTuple
import pandas as pd
import seaborn as sns
import matplotlib as mpl
import scipy.stats as stats
from scipy.stats import mannwhitneyu
import os
from fuzzywuzzy import process
```

```python
# save df to csv
df = pd.read_csv('../logs/results.csv')


graphs_to_keep = ['Karate', 'Dolphins', 'Football', 'Jazz', 'socfb-nips-ego', 'socfb-Mich67', 'socfb-Brandeis99', 'ego-facebook', 'CA-GrQc', 'CA-HepTh']
df = df[df['graphname'].isin(graphs_to_keep)]

```

```python
# Load data from the text file
df2 = pd.read_csv('../logs/resultados1.txt')
# Split the single column by '|' separator
df2_split = df2.iloc[:, 0].str.split('|', expand=True)

# Clean up whitespace and assign proper column names
df2_split.columns = ['graphname', 'bestVND_ILS', 'avgVND_ILS']
df2_split = df2_split.apply(lambda x: x.str.strip())

# Extract numeric values from 'best' and 'average' columns
df2_split['bestVND_ILS'] = df2_split['bestVND_ILS'].str.extract(r'(\d+)').astype(int)
df2_split['avgVND_ILS'] = df2_split['avgVND_ILS'].str.extract(r'([\d.]+)').astype(float)

# Clean graphname (remove .txt extension)
df2_split['graphname'] = df2_split['graphname'].str.replace('.txt', '', regex=False)

# Add missing Karate row
karate_row = pd.DataFrame({'graphname': ['Karate'], 'bestVND_ILS': [3], 'avgVND_ILS': [3.0]})
df2_split = pd.concat([karate_row, df2_split], ignore_index=True)

df2 = df2_split



# Map graphname in df2 to match those in df for merging
name_map = {
    'dolphins': 'Dolphins',
    'football': 'Football',
    'jazz': 'Jazz',
    'ca-GrQc': 'CA-GrQc',
    'ca-HepTh': 'CA-HepTh'
}
df2['graphname'] = df2['graphname'].replace(name_map)

# Merge bestVND_ILS and avgVND_ILS into df
df = df.merge(df2[['graphname', 'bestVND_ILS', 'avgVND_ILS']], on='graphname', how='left')
df.to_csv('../logs/results.csv', index=False)
```
